```python
# Import python Libraries
import pandas as pd
import hvplot.pandas
import holoviews as hv
import osmnx as osm
```

```python
# Search for boundary using OSM

futa_boundary = osm.geocode_to_gdf(

                        "W711995935",

                        by_osmid=True

)

futa_boundary
```

```python
futa_boundary.plot()
```

```python
# This show only the edge boundary with no fill 

futa_boundary.boundary.plot()
```

```python
# Make an interactive plot


futa_map = futa_boundary.hvplot(

    # Givethe map a descriptive title,
    title = "Federal University of Technology Akure, Nigeria",
    
    # Add a basemap
    geo=True, tiles='EsriImagery',
    
    # Change the colors
    fill_color='white', fill_alpha=0.2,
    line_color='skyblue', line_width=5
    
    # Change the image size
   # frame_width=400, frame_height=400 
   )
 

# Display the map 
futa_map
```

```python
# Save the map as a file to put on the web (OPTIONAL)
# Save map as .html file format (it also shows up in my repo on the left)

hv.save(futa_map, 'futa_map2.html')

 

# Display the map

futa_map
```
